```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

```python
df = cc.drop('CUST_ID',axis = 1)
df.head()
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(df)
x_scaled[:5]
```

```python
intertia = []
silhouette = []
K = range(2,11)
for k in K:
    Kmeans = KMeans(n_clusters = k, random_state = 42)
    Kmeans.fit(x_scaled)
    intertia.append(Kmeans.inertia_)
    silhouette.append(silhouette_score(x_scaled,Kmeans.labels_))
```

```python
print(intertia)
print(silhouette)
```

```python
# Plot Elbow
plt.plot(K,intertia,'bx-')
plt.xlabel('k')
plt.ylabel('intertia')
plt.title('Elbow method')
plt.show()
```

```python
plt.plot(K,silhouette,'ro-')
plt.xlabel('k')
plt.ylabel('silhoutte')
plt.title('silhoutte score')
plt.show()
```

```python
# Buliding an algorithum
```

```python
kmeans = KMeans(n_clusters = 7 , random_state = 42 , algorithm = 'elkan')
df['cluster'] = kmeans.fit_predict(x_scaled)
```

```python
print(df['cluster'].value_counts())
```

```python
cluster_profile = df.groupby('cluster').mean()
cluster_profile
```

```python
plt.figure(figsize = (10,4))
sns.scatterplot(x = df['PURCHASES_INSTALLMENTS_FREQUENCY'], y = df['BALANCE_FREQUENCY'],hue = df['cluster'], palette = 'Blues',alpha = 0.8)
plt.title("Customer segementation by balance vs purchase")
plt.show()
```

<h1>Hierarchical Clustring</h1>

```python
cc = pd.read_csv("CC GENERAL.csv")
cc.head()
```

```python
df = cc.drop('CUST_ID',axis = 1)
df.head()
```

```python
df.isna().sum()
```

```python
df['CREDIT_LIMIT'].skew()
```

```python
df['MINIMUM_PAYMENTS'].skew()
```

```python
#Handling null vlues using median imputation
df['CREDIT_LIMIT'] = df['CREDIT_LIMIT'].fillna(df['CREDIT_LIMIT'].median)
df['MINIMUM_PAYMENTS'] = df['MINIMUM_PAYMENTS'].fillna(df['MINIMUM_PAYMENTS'].median)
```

```python
df.isna().sum()
```

```python
obj = df.select_dtypes(include = 'O')
obj.head()
```

```python
X = df.select_dtypes(include=['int64', 'float64'])
```

```python
X[:10]
```

```python
from scipy.cluster.hierarchy import dendrogram , linkage , fcluster
```

```python
z = linkage(X,method ='ward')
```

```python
plt.figure(figsize = (8,4))
dendrogram(z,truncate_mode = 'level',p = 3)
plt.title("Hierarchical clustering")
plt.xlabel("Data points")
plt.ylabel("Distance")
# plt.savefig("Dendorgram")
plt.show()
```

```python
clusters = fcluster(z,3,criterion = 'maxclust')
print(clusters[:10])
```
